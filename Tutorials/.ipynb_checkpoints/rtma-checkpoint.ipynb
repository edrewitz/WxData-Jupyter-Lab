## Downloading the Real Time Mesoscale Analysis (RTMA) Using WxData

In this example, we will use WxData to download the following:

1) Current RTMA for CONUS.
2) 24-Hour RTMA Comparison for Hawaii.

## Importing the `rtma` and `rtma_comparison` functions from WxData

```python
def rtma(model='rtma', 
         cat='analysis',

         process_data=True,

        delete_previous_netcdf_file=True,
        return_values=True):

def rtma_comparison(model='rtma',

```
