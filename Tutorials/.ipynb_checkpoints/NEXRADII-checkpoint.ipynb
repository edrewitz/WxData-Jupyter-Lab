# Downloading NEXRAD II Data

In this example, we will use `download_current_single_station_nexrad2_radar_data()` from WxData to download the latest radar data from KLOT

The WxData NEXRAD II radar clients download the radar data from Amazon AWS Open Data Servers and returns a list of Py-ART radar objects the user can then use to analyze and plot the radar data.

Users can choose how many hours back they wish to download (the default is 3 hours). 

```python

                                                        hours=3,

                                                        proxies=None,
                                                        chunk_size=8192,
                                                        clear_recycle_bin=False,
                                                        mode='precipitation',
                                                        notifications='off'):
```

## Imports

````python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import pyart

from wxdata import download_current_single_station_nexrad2_radar_data
from matplotlib.animation import ArtistAnimation
````

### Calling and using `download_current_single_station_nexrad2_radar_data()`

````python
# We will query KLOT for the past 1 hour of data
radars = download_current_single_station_nexrad2_radar_data('klot', hours=1)
````

````python
radars
````

````python
# We will plot the initial frame radars[0]
display = pyart.graph.RadarDisplay(radars[0])

# Create figure
fig = plt.figure()

# Add subplot
ax = fig.add_subplot(1,1,1)

# Plot radar reflectivity using Py-ART
display.plot("reflectivity", 0, vmin=-32, vmax=64.0)

# Display image
plt.show()
````
