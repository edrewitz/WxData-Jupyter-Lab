## Downloading the NOAA/NWS NDFD Grids (NWS Forecast) Using WxData

In this example, we will use WxData to download the Maximum Temperature Forecast for Hawaii.

WxData solves the issue of var(values) for this data by automatically re-gridding it.

WxData downloads, pre/post-processes and returns a the post-processed xarray.array to the user. 

## Importing the `get_ndfd_grids` function from WxData

```python
def get_ndfd_grids(parameter,

```

[Get NDFD Grids Documentation](https://edrewitz.github.io/WxData/noaa)

````python
from wxdata import get_ndfd_grids
````

````python
ds_short, ds_extended = get_ndfd_grids('maximum_temperature',
                                        'hi')
````
